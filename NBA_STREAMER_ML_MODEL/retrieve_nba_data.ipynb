```python
import time
import pandas as pd
import os
from nba_api.stats.endpoints import leaguegamelog
```

```python
# Define the NBA seasons that will be scraped and used for the model
seasons = ['2020-21', '2021-22', '2022-23', '2023-24', '2024-25', '2025-26']
```

```python
# Directory to store the data
DATA_DIRECTORY = "nba_training_data"

# Directory to store the box scores
SCORES_DIRECTORY = os.path.join(DATA_DIRECTORY, "scores")
os.makedirs(SCORES_DIRECTORY, exist_ok=True)
```

```python
# Function to scrape the NBA website for all the data
def fetch_raw_nba_data(seasons):
        all_season_dfs = []

        for season in seasons:
                print(f"Fetching full player box scores for the {season} season...")
                try:
                        log = leaguegamelog.LeagueGameLog(
                        season=season,
                        player_or_team_abbreviation='P',
                        season_type_all_star='Regular Season'
                        )
                
                        df_season = log.get_data_frames()[0]
                        all_season_dfs.append(df_season)
                        print(f"  -> Success: Extracted {len(df_season)} player game logs.")
                except Exception as e:
                        print(f"  -> Error fetching season {season}: {e}")

                time.sleep(2)

        master_df = pd.concat(all_season_dfs, ignore_index=True)
        print(f"\nDownload complete! Total raw rows fetched: {len(master_df)}")

        output_filename = os.path.join(SCORES_DIRECTORY, "raw_nba_player_boxscores_6yr.csv")
        master_df.to_csv(output_filename, index=False)
        print(f"Raw data saved locally to '{output_filename}'")

        return master_df
```

```python
# Call the function and print a portion of the dataset
df = fetch_raw_nba_data(seasons)

print("\nPreview of the raw dataset:")
print(df[['PLAYER_NAME', 'GAME_DATE', 'MATCHUP', 'PTS', 'REB', 'AST', 'MIN']].head(10))
```
